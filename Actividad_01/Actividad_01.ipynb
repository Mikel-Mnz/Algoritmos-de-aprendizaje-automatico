# Datos del Dataset

## Explicación de los registros
* **Hours_studied:** *Número de horas de estudio por semana*
* **Attendance:** *Porcentaje de asistencia en el periodo*
* **Parental_Involvment:** *Nivel de apoyo académico por parte de los padres*
* **Access_to_Resources:** *Disponibilidad de recursos académicos como libros o internet*
* **Extracurricular_Activities:** *Si el estudiante está involucrado en actividades extracurriculares*
* **Sleep_Hours:** *Promedio de horas de sueño al día*
* **Previous_Scores:** *Promedio de calificaciones académicas previas*
* **Motivation_Level:** *Nivel de motivación del estudiante*
* **Internet_Access:** *Si el estudiante tiene acceso al internet*
* **Tutoring_Sessions:** *Número de sesiones de tutoría a la semana*
* **Family_Income:** *Nivel de ingresos familiares*
* **Teacher_Quality:** *Calidad general percibida de los docentes*
* **School_Type:** *Tipo de escuela (Pública o Privada)*
* **Peer_Influence:** *Influencia de los compañeros en el rendimiento académico*
* **Physical_Activity:** *Horas de actividad física a la semana*
* **Learning_Disabilities:** *Si el estudiante presenta dificultades de aprendizaje*
* **Parental_Education_Level:** *Nivel máximo de estudios de los padres*
* **Distance_from_Home:** *Distancia entre el hogar y la escuela (Cerca, Moderada, Lejos)*
* **Gender:** *Género del estudiante*
* **Exam_Score:** *Calificación final del examen*

### Variable objetivo
La variable **Exam_Score** es nuestra variable objetivo.

### Tipo de problema de aprendizaje automático
Consideramos que este caso representaría un tipo de aprendizaje automático **supervisado** ya que se intenta predecir y recomendar cuál o cuáles son los factores principales para que un estudiante pueda obtener una mejor calificación en su examen a partir de los datos de este csv.

### Utilidad del modelo
Con estos datos se podría construir un modelo el cual nos permita conocer y predecir cuáles son los principales factores que favorecen a los estudiantes a obtener una mayor calificación con un examen. De esta forma, los estudiantes pueden enfocarse mejor en esos factores y dejar de preouparse por los otros. Al mismo tiempo, este modelo ayudaría a las instituciones educativas en proveer mejores herramientas a los estudiantes para que tengan un mejor desempeño académico.
En conclusión, este modelo ayudaría a tomar mejores decisiones a los estudiantes, escuelas y familias con el objetivo de obtener una mejor puntuación en su examen.

| Problema detectado | Variable afectada | Evidencia | Tratamiento propuesto |
| :--- | :--- | :--- | :--- |
| **Valores faltantes (Nulos)** | `Teacher_Quality`<br>`Parental_Education_Level`<br>`Distance_from_Home` | Al utilizar el método de la libreria de Pandas para mostrar la suma de valores nulos por variable, podemos observar 78, 90 y 67 registros nulos en las variables ya mencionadas. | Al ser variables de tipo categórico y solamente representar entre el 1% y el 1.3% de los datos dentro del Dataset, tenemos dos opciones que podemos seguir, la primera es llenar esos valores nulos con la moda y la segunda es eliminar esos registros. |
| **Valores imposibles** | `Exam_Score` | En el meétodo de la libreria de python que muestra las estadísticas descriptivas, el valor máximo (`max`) de esa variable es de **101**. La calificación normal de un examen típicamente no es mayor a 100. | Buscar el número de registros con ese valor atípico y actualizar su valor a 100, al probablemente ser un error tipográfico. |

```python
moda_teacher_quality = df["Teacher_Quality"].mode()[0]
print(moda_teacher_quality)
```

```python
df["Teacher_Quality"] = df["Teacher_Quality"].fillna(moda_teacher_quality)
print(df["Teacher_Quality"].isna().sum())
```

```python
mayores_100 = (df['Exam_Score'] > 100).sum()
print(f"Registros con calificación mayor a 100: {mayores_100}")

menores_0 = (df['Exam_Score'] < 0).sum()
print(f"Registros con calificación menor a 0: {menores_0}")
```

```python
df.loc[df['Exam_Score'] > 100, 'Exam_Score'] = 100

print("Valor máximo de Exam_Score : ", df['Exam_Score'].max())
```

```python
#Detección de valores atípicos en la columna "Hours_Studied"
Q1 = df["Hours_Studied"].quantile(0.25)
Q3 = df["Hours_Studied"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = df[(df["Hours_Studied"] < lower_limit) | (df["Hours_Studied"] > upper_limit)]

print(f"Número de valores atípicos en 'Hours_Studied': {len(outliers)}")
print(outliers[["Hours_Studied"]].head())

#Estadísticas descriptivas de la columna "Hours_Studied"
print("\nEstadísticas descriptivas de la columna 'Hours_Studied':")
print(df["Hours_Studied"].describe())

#Histograma de la columna "Hours_Studied"
plt.figure(figsize=(8, 6))
sns.histplot(df["Hours_Studied"], bins=20, kde=True, color='blue')
plt.title("Distribución de Horas Estudiadas")
plt.xlabel("Horas Estudiadas")
plt.ylabel("Frecuencia")
plt.show()

#Boxplot de la columna "Hours_Studied"
plt.figure(figsize=(8, 6))
sns.boxplot(x=df["Hours_Studied"], color='lightblue')
plt.title("Boxplot de Horas Estudiadas")
plt.xlabel("Horas Estudiadas")
plt.show()
```
