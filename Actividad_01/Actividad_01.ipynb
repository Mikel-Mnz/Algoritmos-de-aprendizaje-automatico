# Datos del Dataset

## Explicación de los registros
* **Hours_studied:** *Número de horas de estudio por semana*
* **Attendance:** *Porcentaje de asistencia en el periodo*
* **Parental_Involvment:** *Nivel de apoyo académico por parte de los padres*
* **Access_to_Resources:** *Disponibilidad de recursos académicos como libros o internet*
* **Extracurricular_Activities:** *Si el estudiante está involucrado en actividades extracurriculares*
* **Sleep_Hours:** *Promedio de horas de sueño al día*
* **Previous_Scores:** *Promedio de calificaciones académicas previas*
* **Motivation_Level:** *Nivel de motivación del estudiante*
* **Internet_Access:** *Si el estudiante tiene acceso al internet*
* **Tutoring_Sessions:** *Número de sesiones de tutoría a la semana*
* **Family_Income:** *Nivel de ingresos familiares*
* **Teacher_Quality:** *Calidad general percibida de los docentes*
* **School_Type:** *Tipo de escuela (Pública o Privada)*
* **Peer_Influence:** *Influencia de los compañeros en el rendimiento académico*
* **Physical_Activity:** *Horas de actividad física a la semana*
* **Learning_Disabilities:** *Si el estudiante presenta dificultades de aprendizaje*
* **Parental_Education_Level:** *Nivel máximo de estudios de los padres*
* **Distance_from_Home:** *Distancia entre el hogar y la escuela (Cerca, Moderada, Lejos)*
* **Gender:** *Género del estudiante*
* **Exam_Score:** *Calificación final del examen*

### Variable objetivo
La variable **Exam_Score** es nuestra variable objetivo.

### Tipo de problema de aprendizaje automático
Consideramos que este caso representaría un tipo de aprendizaje automático **supervisado** ya que se intenta predecir y recomendar cuál o cuáles son los factores principales para que un estudiante pueda obtener una mejor calificación en su examen a partir de los datos de este csv.

### Utilidad del modelo
Con estos datos se podría construir un modelo el cual nos permita conocer y predecir cuáles son los principales factores que favorecen a los estudiantes a obtener una mayor calificación con un examen. De esta forma, los estudiantes pueden enfocarse mejor en esos factores y dejar de preouparse por los otros. Al mismo tiempo, este modelo ayudaría a las instituciones educativas en proveer mejores herramientas a los estudiantes para que tengan un mejor desempeño académico.
En conclusión, este modelo ayudaría a tomar mejores decisiones a los estudiantes, escuelas y familias con el objetivo de obtener una mejor puntuación en su examen.

| Problema detectado | Variable afectada | Evidencia | Tratamiento propuesto |

# Paso 4

En esta ocasión, al representar poco más del 1% de nuestros datos y siguiendo las buenas prácticas de análisis y limpieza, hemos optado por mantener los registros con valores nulos y, en su lugar, llenarlos usando el valor que más se repite: la moda.

Al ser muy pocos los valores nulos, no son suficientes como para generar un sesgo y afectar los resultados de nuestros datos.

```python
moda_teacher_quality = df["Teacher_Quality"].mode()[0]
print(moda_teacher_quality)
```

```python
df["Teacher_Quality"] = df["Teacher_Quality"].fillna(moda_teacher_quality)
print(df["Teacher_Quality"].isna().sum())
```

```python
moda_parental_education = df["Parental_Education_Level"].mode()[0]
print(f"Moda de Parental_Education_Level: {moda_parental_education}")
```

En el caso del valor atípico encontradp en 'Exam_Score' (un puntaje de 101), se concluyó que se trata de un error tipográfico, ya que la escala normal de evaluación de un examen es de 0 a 100. 

La evidencia en la que basamos esta afirmación es que el límite máximo en la columna 'Previous_Scores' también es 100. La solución aplicada consistió en localizar el registro y ajustar el valor atípico al límite máximo permitido (100)."

```python
mayores_100 = (df['Exam_Score'] > 100).sum()
print(f"Registros con calificación mayor a 100: {mayores_100}")

menores_0 = (df['Exam_Score'] < 0).sum()
print(f"Registros con calificación menor a 0: {menores_0}")
```

```python
df.loc[df['Exam_Score'] > 100, 'Exam_Score'] = 100

print("Valor máximo de Exam_Score : ", df['Exam_Score'].max())
```

# Paso 5

```python
# Descripciones
mean_val = df['Exam_Score'].mean()
median_val = df['Exam_Score'].median()
std_val = df['Exam_Score'].std()
skew_val = df['Exam_Score'].skew()

print("--- Medidas Descriptivas ---")
print(df['Exam_Score'].describe())
print(f"Sesgo (Skewness): {skew_val:.2f}")

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Histograma
sns.histplot(df['Exam_Score'], kde=True, ax=axes[0], color='skyblue')
axes[0].set_title('Histograma de Exam_Score')
axes[0].axvline(mean_val, color='red', linestyle='--', label=f'Media: {mean_val:.2f}')
axes[0].axvline(median_val, color='green', linestyle='-', label=f'Mediana: {median_val:.2f}')
axes[0].legend()

# Boxplot
sns.boxplot(x=df['Exam_Score'], ax=axes[1], color='lightgreen')
axes[1].set_title('Diagrama de Caja de Exam_Score')

plt.tight_layout()
plt.show()
```

1. Interpretación de la distribución

Comportamiento: Aunque la media (67.24) y la mediana (67.00) coinciden en el centro, el sesgo de 1.64 confirma que la distribución está asimétrica y desplazada hacia la derecha.

Concentración: El 50% central de los estudiantes se agrupa en un rango muy estrecho entre 65 y 69 puntos (los cuartiles Q1 y Q3).

2. Identificación de valores atípicos

Presencia de atípicos: El boxplot revela una gran concentración de valores atípicos en el extremo superior (por encima de 75 puntos) y unos pocos en el extremo inferior (por debajo de 59).

Dato anómalo claro: El valor máximo registrado es 101.00, el cual representa una inconsistencia o error de captura si la escala de evaluación es de 0 a 100.

3. Problemas potenciales para el modelado posterior

Limpieza de datos requerida: El registro con calificación de 101.00 debe corregirse o eliminarse para evitar distorsionar el entrenamiento.

Sensibilidad a algoritmos: Los atípicos extremos superiores pueden sesgar negativamente modelos lineales (Regresión Lineal) y algoritmos basados en distancias (K-Means).

Preprocesamiento recomendado: Dada la asimetría y la cantidad de atípicos, no es conveniente usar StandardScaler. Se recomienda usar RobustScaler o aplicar una transformación logarítmica a la variable objetivo.

```python
df_encoded = df.copy()
```

```python
# Mapeo explícito para preservar el orden jerárquico natural
map_parental_involvement = {
    'Low': 0,
    'Medium': 1,
    'High': 2
}

map_motivation_level = {
    'Low': 0,
    'Medium': 1,
    'High': 2
}

# Aplicación del mapeo
df_encoded['Parental_Involvement_Encoded'] = df_encoded['Parental_Involvement'].map(map_parental_involvement)
df_encoded['Motivation_Level_Encoded'] = df_encoded['Motivation_Level'].map(map_motivation_level)

print("--- Mapeo Ordinal Aplicado ---")
print(df_encoded[['Parental_Involvement', 'Parental_Involvement_Encoded']].drop_duplicates())
print("\n", df_encoded[['Motivation_Level', 'Motivation_Level_Encoded']].drop_duplicates())
```

```python
# Aplicación de One-Hot Encoding a variables nominales
# drop_first=True evita la colinealidad perfecta 
df_encoded = pd.get_dummies(
    df_encoded, 
    columns=['School_Type', 'Internet_Access'], 
    drop_first=False, 
    dtype=int
)

# Visualización del resultado
columnas_nuevas = [col for col in df_encoded.columns if 'School_Type' in col or 'Internet_Access' in col or '_Encoded' in col]
print("--- Columnas Transformadas (Muestra de 5 registros) ---")
print(df_encoded[columnas_nuevas].head())
```

Diferencia entre categorías nominales y variables con orden 

Variables Ordinales tienen orden: Poseen una jerarquía o nivel intrínseco.

Variables Nominales no tienen orden: Son etiquetas descriptivas sin ninguna relación de mayor/menor ni jerarquía.

¿Por qué se eligió cada método?

Codificación Ordinal para Parental_Involvement y Motivation_Level:

Por qué se eligió: Se asignan números enteros crecientes para reflejar matemáticamente la jerarquía. Esto permite que el modelo entienda que High representa un nivel superior a Medium y Low.

One-Hot Encoding para School_Type e Internet_Access:

Por qué se eligió: Convierte cada categoría en una columna binaria independiente. Si usáramos números como 0 y 1 en una sola columna con categorías múltiples sin orden, el algoritmo asumiría erróneamente que una opción vale más o está "más lejos" que otra. One-Hot Encoding mantiene todas las opciones en igualdad de condiciones algebraicas.

```python
# 1. Instanciar el escalador
scaler = StandardScaler()

# 2. Variables seleccionadas para el escalamiento
cols_to_scale = ['Hours_Studied', 'Attendance']

# 3. Ajustar y transformar
df_scaled = df_encoded.copy()
df_scaled[[f'{col}_Scaled' for col in cols_to_scale]] = scaler.fit_transform(df_scaled[cols_to_scale])
```

```python
# Comparación de estadísticas descriptivas
print("--- Estadísticas Originales ---")
print(df_scaled[cols_to_scale].describe().round(3))

print("\n--- Estadísticas Después de StandardScaler (Media ≈ 0, Desv. Estándar = 1) ---")
print(df_scaled[[f'{col}_Scaled' for col in cols_to_scale]].describe().round(3))

print("\n--- Muestra de Registros Antes y Después ---")
print(df_scaled[['Hours_Studied', 'Hours_Studied_Scaled', 'Attendance', 'Attendance_Scaled']].head())
```

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Hours_Studied - Original
sns.histplot(df_scaled['Hours_Studied'], kde=True, ax=axes[0, 0], color='royalblue')
axes[0, 0].set_title('Hours_Studied (Original: Escala en Horas)')
axes[0, 0].set_xlabel('Horas')

# Hours_Studied - Escalado
sns.histplot(df_scaled['Hours_Studied_Scaled'], kde=True, ax=axes[0, 1], color='crimson')
axes[0, 1].set_title('Hours_Studied (StandardScaler: Centrado en 0, std=1)')
axes[0, 1].set_xlabel('Z-Score')

# Attendance - Original
sns.histplot(df_scaled['Attendance'], kde=True, ax=axes[1, 0], color='royalblue')
axes[1, 0].set_title('Attendance (Original: Escala en Porcentaje 0-100)')
axes[1, 0].set_xlabel('Porcentaje')

# Attendance - Escalado
sns.histplot(df_scaled['Attendance_Scaled'], kde=True, ax=axes[1, 1], color='crimson')
axes[1, 1].set_title('Attendance (StandardScaler: Centrado en 0, std=1)')
axes[1, 1].set_xlabel('Z-Score')

plt.tight_layout()
plt.show()
```

### Comparación de valores
**Antes:** Las variables conservan sus unidades reales. Hours_Studied va de ~1 a ~40 horas con media en ~20, mientras que Attendance va de ~60% a ~100% con media en ~80%.

**Después:** Ambas variables se transforman a puntuaciones $Z$ sin unidades: su media se vuelve exactamente 0 y su desviación estándar se vuelve 1, quedando la mayoría de los datos concentrados en un rango común entre $-3$ y $3$.

### ¿Qué cambia y qué información se conserva?

**Lo que cambia:** Las unidades de medida originales, la magnitud absoluta de los números y el punto central de los datos (la media).

**Lo que se conserva:** La forma de la distribución, las posiciones relativas  y las relaciones entre las variables.

### ¿Por qué algunos algoritmos son sensibles a la escala?

**Modelos basados en distancias:** Miden qué tan "cerca" están dos observaciones en el espacio. Si una variable tiene números grandes  y otra números pequeños, la variable grande domina la distancia y el modelo ignora la pequeña.

**Modelos optimizados por gradiente:** Actualizan sus pesos paso a paso. Las diferencias de escala deforman el mapa de error, haciendo que el descenso de gradiente oscile en zigzag y tarde mucho más en aprender.

**Modelos regularizados:** Castigan coeficientes altos para evitar sobreajuste. Al no escalar, penalizan con mayor dureza a las variables con rangos numéricos pequeños debido a la magnitud que toman sus pesos.

**Nota importante:** en una etapa posterior de modelado, tanto la codificación ordinal 
como el escalamiento (`StandardScaler`) deberán ajustarse (`fit`) únicamente con los 
datos de entrenamiento, y luego aplicarse (`transform`) tanto al conjunto de entrenamiento

Para asegurar la integridad de los datos, se aplicaron transformaciones correctivas directas. Asimismo, se realizó la imputación de los valores faltantes en la variable `Teacher_Quality` utilizando la moda (valor más frecuente) para no alterar la distribución de la variable.

**3. Efecto de la codificación y el escalamiento**
La transformación de las variables permitió adaptar el dataset para que sea procesable por algoritmos de Machine Learning:
*   **Codificación:** El mapeo ordinal aplicado a `Parental_Involvement` y `Motivation_Level` logró convertir categorías de texto a números preservando su jerarquía natural (Low < Medium < High). Antes de alimentar un modelo predictivo, es estrictamente necesario:
