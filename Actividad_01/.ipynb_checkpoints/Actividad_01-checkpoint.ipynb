# Datos del Dataset

## Explicación de los registros
* **Hours_studied:** *Número de horas de estudio por semana*
* **Attendance:** *Porcentaje de asistencia en el periodo*
* **Parental_Involvment:** *Nivel de apoyo académico por parte de los padres*
* **Access_to_Resources:** *Disponibilidad de recursos académicos como libros o internet*
* **Extracurricular_Activities:** *Si el estudiante está involucrado en actividades extracurriculares*
* **Sleep_Hours:** *Promedio de horas de sueño al día*
* **Previous_Scores:** *Promedio de calificaciones académicas previas*
* **Motivation_Level:** *Nivel de motivación del estudiante*
* **Internet_Access:** *Si el estudiante tiene acceso al internet*
* **Tutoring_Sessions:** *Número de sesiones de tutoría a la semana*
* **Family_Income:** *Nivel de ingresos familiares*
* **Teacher_Quality:** *Calidad general percibida de los docentes*
* **School_Type:** *Tipo de escuela (Pública o Privada)*
* **Peer_Influence:** *Influencia de los compañeros en el rendimiento académico*
* **Physical_Activity:** *Horas de actividad física a la semana*
* **Learning_Disabilities:** *Si el estudiante presenta dificultades de aprendizaje*
* **Parental_Education_Level:** *Nivel máximo de estudios de los padres*
* **Distance_from_Home:** *Distancia entre el hogar y la escuela (Cerca, Moderada, Lejos)*
* **Gender:** *Género del estudiante*
* **Exam_Score:** *Calificación final del examen*

### Variable objetivo
La variable **Exam_Score** es nuestra variable objetivo.

### Tipo de problema de aprendizaje automático
Consideramos que este caso representaría un tipo de aprendizaje automático **supervisado** ya que se intenta predecir y recomendar cuál o cuáles son los factores principales para que un estudiante pueda obtener una mejor calificación en su examen a partir de los datos de este csv.

### Utilidad del modelo
Con estos datos se podría construir un modelo el cual nos permita conocer y predecir cuáles son los principales factores que favorecen a los estudiantes a obtener una mayor calificación con un examen. De esta forma, los estudiantes pueden enfocarse mejor en esos factores y dejar de preouparse por los otros. Al mismo tiempo, este modelo ayudaría a las instituciones educativas en proveer mejores herramientas a los estudiantes para que tengan un mejor desempeño académico.
En conclusión, este modelo ayudaría a tomar mejores decisiones a los estudiantes, escuelas y familias con el objetivo de obtener una mejor puntuación en su examen.

# Paso 5

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('StudentPerformanceFactors.csv') 

# Descripciones
mean_val = df['Exam_Score'].mean()
median_val = df['Exam_Score'].median()
std_val = df['Exam_Score'].std()
skew_val = df['Exam_Score'].skew()

print("--- Medidas Descriptivas ---")
print(df['Exam_Score'].describe())
print(f"Sesgo (Skewness): {skew_val:.2f}")

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Histograma
sns.histplot(df['Exam_Score'], kde=True, ax=axes[0], color='skyblue')
axes[0].set_title('Histograma de Exam_Score')
axes[0].axvline(mean_val, color='red', linestyle='--', label=f'Media: {mean_val:.2f}')
axes[0].axvline(median_val, color='green', linestyle='-', label=f'Mediana: {median_val:.2f}')
axes[0].legend()

# Boxplot
sns.boxplot(x=df['Exam_Score'], ax=axes[1], color='lightgreen')
axes[1].set_title('Diagrama de Caja de Exam_Score')

plt.tight_layout()
plt.show()
```

```python
1 Interpretación de la distribución

Comportamiento: Aunque la media(67.24) y la mediana(67.00) coinciden en el centro, el sesgo de 1.64 confirma que la distribución está asimetrica y desplazada hacia la derecha

Concentración: El 50% central de los estudiantes se agrupa en un rango muy estrecho entre 65 y 69 puntos(los cuartiles Q1 y Q3)

2 Identificación de valores atípicos

Presencia de atipicos: El boxplot revela una gran concentración de valores atípicos en el extremo superior (por encima de 75 puntos) y unos pocos en el extremo inferior (por debajo de 59)

Dato anomalo claro: El valor máximo registrado es 101.00, el cual representa una inconsistencia o error de captura si la escala de evaluación es de 0 a 100

3 Problemas potenciales para el modelado posterior

Limpieza de datos requerida: El registro con calificación de 101.00 debe corregirse o eliminarse para evitar distorsionar el entrenamiento

Sensibilidad a algoritmos: Los atipicos extremos superiores pueden sesgar negativamente modelos lineales(Regresion lineal) y algoritmos basados en distancias (K-Means)

Preprocesamiento recomandado: Dada la asimetria y la cantidad de atípicos, no es conveniente usar "StandardScaler". Se recomienda usar "RobustScaler" o aplicar una transformación logarítmica a la variable objetivo
```

# Paso 6

```python
Q1 = df["Hours_Studied"].quantile(0.25)
Q3 = df["Hours_Studied"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = df[(df["Hours_Studied"] < lower_limit) | (df["Hours_Studied"] > upper_limit)]

print(f"Número de valores atípicos en 'Hours_Studied': {len(outliers)}")
print(outliers[["Hours_Studied"]].head())

#Histograma de la columna "Hours_Studied"
plt.figure(figsize=(8, 6))
sns.histplot(df["Hours_Studied"], bins=20, kde=True, color='blue')
plt.title("Distribución de Horas Estudiadas")
plt.xlabel("Horas Estudiadas")
plt.ylabel("Frecuencia")
plt.show()
```

```python
Q1 = df["Sleep_Hours"].quantile(0.25)
Q3 = df["Sleep_Hours"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = df[(df["Sleep_Hours"] < lower_limit) | (df["Sleep_Hours"] > upper_limit)]

print(f"Número de valores atípicos en 'Sleep_Hours': {len(outliers)}")
print(outliers[["Sleep_Hours"]].head())

#Histograma de la columna "Sleep_Hours"
plt.figure(figsize=(8, 6))
sns.histplot(df["Sleep_Hours"], bins=7, kde=True, color='blue')
plt.title("Distribución de Horas de Sueño")
plt.xlabel("Horas de Sueño")
plt.ylabel("Frecuencia")
plt.show()
```

```python
Q1 = df["Attendance"].quantile(0.25)
Q3 = df["Attendance"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = df[(df["Attendance"] < lower_limit) | (df["Attendance"] > upper_limit)]

print(f"Número de valores atípicos en 'Attendance': {len(outliers)}")
print(outliers[["Attendance"]].head())

#Histograma de la columna "Attendance"
plt.figure(figsize=(8, 6))
sns.histplot(df["Attendance"], bins=20, kde=True, color='blue')
plt.title("Distribución de Asistencia")
plt.xlabel("Asistencia")
plt.ylabel("Frecuencia")
plt.show()
```
