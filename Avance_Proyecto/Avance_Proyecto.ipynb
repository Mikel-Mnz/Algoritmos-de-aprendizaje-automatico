# Paso 3: Definición del problema

**Contexto organizacional** 

*Una empresa que ofrece servicios por suscripción ha detectado una disminución en la permanencia de sus clientes. Las acciones de retención tomada se aplican de manera general, por lo que no hay una distinción sobre cuáles clientes presentan un mayor riesgo. La empresa solicita una solución de aprendizaje automático que permita estimar la probabilidad de abandono y generar evidencia para priorizar acciones de seguimiento.*

**Unidad de análisis**

*Cada fila del dataset representa a un cliente único de la compañía. En ellas encontraremos información demográfica, los servicios específicos que tiene contratados y los datos de su cuenta y facturación.*

*A continuación las variables junto con su respectiva y breve descripción:*
* **customerID:** *Identificador único de los clientes*
* **gender:** *Género con el que se identifica a los clientes (Male/Female)*
* **SeniorCitizen:** *Si el cliente es una persona de la tercera edad (0 para falso y 1 para verdadero)*
* **Partner:** *Si el cliente tiene una pareja (Yes/No)*
* **Dependents:** *Si el cliente tiene personas que requieran de un sostén económico (Yes/No)*
* **tenure:** *Número de meses que el cliente ha estado suscrito*
* **PhoneService:** *Si el cliente tiene una línea telefónica (Yes/No)*
* **MultipleLines:** *Si el cliente tiene múltiples líneas telefónicas (Yes/No/No phone service)*
* **InternetService:** *Proveedor de servicio de Internet del cliente (DSL/Fiber optic/No)*
* **OnlineSecurity:** *Si el cliente tiene plan de seguridad en línea (Yes/No/No internet service)*
* **OnlineBackup:** *Si el cliente tiene un plan respaldo en línea (Yes/No/No internet service)*
* **DeciveProtection:** *Si el cliente tiene un plan de protección de dispositivos (Yes/No/No internet service)*
* **TechSupport:** *Si el cliente tiene soporte técnico (Yes/No/No internet service)*
* **StreamingTV:** *Si el cliente tiene acceso a hacer streaming de TV (Yes/No/No internet service)*
* **StreamingMovies:** *Si el cliente tiene acceso a hacer streaming de películas (Yes/No/No internet service)*
* **Contract:** *El tipo de periodo del contrato del cliente (Month-to-month/One year/Two year)*
* **PaperlessBilling:** *Si el cliente tiene facturación electrónica/sin papel (Yes/No)*
* **PaymentMethod:** *El método de pago del cliente (Electronic check/Mailed check/Bank transfer (automatic)/Credit card (automatic))*
* **MonthlyCharges:** *La cantidad mensual cobrada al cliente*
* **TotalCharges:** *La cantidad total cobrada al cliente*
* **Churn:** *Si el cliente abandonó el servicio (Yes/No)*


**Variable objetivo:** 

*La variable objetivo en este caso es **Churn**, la cual indica abandono o permanencia. Se debe entender el Churn como la pérdida de clientes, por lo que un valor verdadero (True/Yes) significa que abandonó la suscripción, mientras que un valor falso (False/No) indica que permaneció con la suscripción.*

**Tipo de problema:** 

*El tipo de problema es una clasificación binaria ya que el objetivo de la empresa es determinar los factores que influyen en el abandono o permanencia del servicio por parte de los clientes. Al final, lo que busca la empresa es saber si el cliente se va o se queda.*

**Clase positiva:** 

*En la variable **Churn**, un valor verdadero significa que el cliente abandonó el servicio.*

**Usuarios potenciales:** 

*Los perfiles que podrían beneficiarse de esto son claramente los departamentos de marketing y finanzas, así como la propia compañía. Un análisis descriptivo y la elaboración de modelos predictivos ayudaría a la compañía a tomar acciones importantes para aumentar la permanencia de los clientes de los servicios que ofrece.* 

**Decisión apoyada:** 

*Con la predicción del modelo a elaborar, se estarían implementando decisiones para optimizar la permanencia de los clientes como lo podrían ser la dirección de campañas de retención personalizadas, la oferta de descuentos o el seguimiento proactivo y prioritario del equipo de atención a clientes a aquellos usuarios con alta probabilidad de abandono.*

**Riesgo de uso incorrecto:** 

*La predicción no debe tomarse como una certeza, sino como una guía para poder ayudar en la toma de decisiones. El tomar esta predicción y realizar acciones de forma automática con base en los resultados del modelo podría llevar a la toma de decisiones erróneas como una terminación de contrato injustificada, una reducción de los servicios suscritos o un aumento desconsiderable en la cantidad del precio de forma que podría afectar a clientes que incluso ni tenían pensado en abandonar el servicio.*
