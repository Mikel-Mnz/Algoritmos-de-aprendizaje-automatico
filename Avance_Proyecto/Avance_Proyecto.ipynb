# Paso 3: Definición del problema

**Contexto organizacional** 

*Una empresa que ofrece servicios por suscripción ha detectado una disminución en la permanencia de sus clientes. Las acciones de retención tomada se aplican de manera general, por lo que no hay una distinción sobre cuáles clientes presentan un mayor riesgo. La empresa solicita una solución de aprendizaje automático que permita estimar la probabilidad de abandono y generar evidencia para priorizar acciones de seguimiento.*

**Unidad de análisis**

*Cada fila del dataset representa a un cliente único de la compañía. En ellas encontraremos información demográfica, los servicios específicos que tiene contratados y los datos de su cuenta y facturación.*

**Variable objetivo:** 

*La variable objetivo en este caso es **Churn**, la cual indica abandono o permanencia. Se debe entender el Churn como la pérdida de clientes, por lo que un valor verdadero (True/Yes) significa que abandonó la suscripción, mientras que un valor falso (False/No) indica que permaneció con la suscripción.*

**Tipo de problema:** 

*El tipo de problema es una clasificación binaria ya que el objetivo de la empresa es determinar los factores que influyen en el abandono o permanencia del servicio por parte de los clientes. Al final, lo que busca la empresa es saber si el cliente se va o se queda.*

**Clase positiva:** 

*En la variable **Churn**, un valor verdadero significa que el cliente abandonó el servicio.*

**Usuarios potenciales:** 

*Los perfiles que podrían beneficiarse de esto son claramente los departamentos de marketing y finanzas, así como la propia compañía. Un análisis descriptivo y la elaboración de modelos predictivos ayudaría a la compañía a tomar acciones importantes para aumentar la permanencia de los clientes de los servicios que ofrece.* 

**Decisión apoyada:** 

*Con la predicción del modelo a elaborar, se estarían implementando decisiones para optimizar la permanencia de los clientes como lo podrían ser la dirección de campañas de retención personalizadas, la oferta de descuentos o el seguimiento proactivo y prioritario del equipo de atención a clientes a aquellos usuarios con alta probabilidad de abandono.*

**Riesgo de uso incorrecto:** 

*La predicción no debe tomarse como una certeza, sino como una guía para poder ayudar en la toma de decisiones. El tomar esta predicción y realizar acciones de forma automática con base en los resultados del modelo podría llevar a la toma de decisiones erróneas como una terminación de contrato injustificada, una reducción de los servicios suscritos o un aumento desconsiderable en la cantidad del precio de forma que podría afectar a clientes que incluso ni tenían pensado en abandonar el servicio.*

**Criterio inicial de éxito:** 

*El criterio inicial de éxito se podría ver reflejado en la disminución considerable de clientes que abandonaron el servicio y, por lo tanto, en el aumento de clientes que decidieron permanecer.*

# Paso 9 Revisión de desbalance

```python
proporcion_clases = df['Churn'].value_counts(normalize=True) * 100

df_desbalance = pd.DataFrame({
    'Proporción (%)': proporcion_clases.round(2)
})

print("Proporción de clases en la variable objetivo 'Churn':")
display(df_desbalance)
```

- Al observar la proporción de la variable objetivo, existe un claro desbalance de clases (73.46% para la clase mayoritaria “No” frente a un 26.54% para la clase minoritaria "Yes"). El efecto principal de esto es que, durante el entrenamiento, un modelo  podría sesgar hacia la clase mayoritaria para minimizar el error global, dándonos una falsa sensación de que el modelo tiene un gran rendimiento, cuando en realidad solo está asumiendo que casi ningún cliente abandona el servicio.

- En un dataset donde el 73.46% de los clientes se queda, un modelo que simplemente prediga "No" para absolutamente todos los casos obtendría un 73.46% de accuracy. Aunque este número parece aceptable en papel, el modelo sería completamente inútil en la práctica porque fallaría en detectar al 100% de los clientes que sí se van.

- Debido a este desbalance, la exactitud no es confiable. Debemos enfocarnos en métricas derivadas de la matriz de confusión:

    1.-Recall (Sensibilidad): Es la métrica más crítica en este proyecto, ya que mide qué porcentaje de los clientes que realmente abandonaron logramos identificar a tiempo.

    2.-Precisión: Nos indica cuántos de los clientes etiquetados por el modelo como "en riesgo de abandono" realmente lo estaban, ayudando a no desperdiciar recursos en falsas alarmas.

    3.-F1-Score: Proporciona un equilibrio armónico entre Precisión y Recall.

- Si el modelo genera muchos "Falsos Negativos" (clientes que abandonan pero el modelo predijo que se quedarían), la empresa pierde la oportunidad de intervenir. Al no estar etiquetados como riesgo, el equipo de retención no les ofrecerá descuentos, negociaciones ni seguimiento proactivo, traduciéndose en una pérdida directa e irrecuperable de clientes e ingresos.

- En esta etapa de diseño y exploración inicial, es fundamental entrenar primero un modelo con los datos en su estado natural para conocer su rendimiento real. Además, muchos algoritmos pueden manejar este desbalance ajustando los pesos de las clases internamente de forma matemática, sin necesidad de alterar los registros del dataset.

- En caso de que en iteraciones futuras se decida utilizar técnicas de remuestreo para intentar mejorar el Recall, estas deberán aplicarse única y exclusivamente sobre los datos de entrenamiento.
Por ningún motivo se deberá aplicar remuestreo sobre el conjunto completo ni sobre el conjunto de prueba. Hacerlo provocaría una grave fuga de información, evaluando al modelo en un escenario ficticio. El conjunto de prueba siempre debe conservar el desbalance del 73.46%/26.54% original para medir honestamente cómo se comportará el modelo en producción.

```python
# Convertir TotalCharges a numérico primero (algunos valores vienen como espacio en blanco)
df['TotalCharges_calc'] = pd.to_numeric(df['TotalCharges'], errors='coerce')

print("Cantidad de valores no convertibles en TotalCharges:")
print(df['TotalCharges_calc'].isnull().sum())

print("\nCorrelación entre tenure, MonthlyCharges y TotalCharges:")
print(df[['tenure', 'MonthlyCharges', 'TotalCharges_calc']].corr())

print("\nDistribución de tenure según Churn:")
print(df.groupby('Churn')['tenure'].describe())

# Verificar si TotalCharges es aproximadamente tenure * MonthlyCharges
df['diff_TotalCharges'] = (df['TotalCharges_calc'] - df['tenure'] * df['MonthlyCharges']).abs()
print("\nDiferencia entre TotalCharges real y tenure*MonthlyCharges calculado:")
print(df['diff_TotalCharges'].describe())
```

No se identificaron variables que contengan información generada despues del evento de abandono(no existen campos como fecha de cancelación o motivo de baja), por lo que no hay una fuga de información directa en el dataset.

Sin embargo, se detectó un riesgo de fuga indirecta y conceptual en la variable "tenure". La evidencia lo confirma los clientes que abandonaron (CHUrn = Yes) tienen una antigüedad promedio de 17.98 meses, notablemente menor a los 37.57 meses de los clientes que permanecieron. Esto se debe a que, para los clientes que se dieron de baja, "tenure" representa literalmente "cuánto tardaron en abandonar", quedando congelado en ese punto - no es un dato que evolucione de forma independiente al evento que se busca predecir.

"TotalCharges" presenta una correlación alta con "tenure" (0.826) y moderada con "MonthlyCharges" (0.651), lo que confirma que hereda parcialmente ese mismo riesgo conceptual, aunque no es una copia exacta: la diferencia entre "TotalCharges" real y el cálcula "tenure x MontlhyCharges" tiene una media de 45.09 (con máximo de 373.25), lo que indica que existen otros factores (cambios de plan, promociones, cargos variables en el tiempo) que también influyen en el total facturado.

tenure

Cuándo se genera: Se actualiza mes a mes mientras el cliente está activo. Para clientes que abandonaron, el valor queda fijo en el mes exacto en que se dieron de baja.

¿Disponible al momento de predecir? Sí, siempre se puede calcular "meses transcurridos a la fecha" — el problema no es de disponibilidad, sino conceptual.

¿Podría inflar métricas artificialmente? Sí, de forma indirecta: los clientes con Churn = Yes tienen tenure promedio de 17.98 meses vs. 37.57 de los que permanecieron — el modelo aprendería a asociar tenure bajo con churn parcialmente por cómo se construyó el dataset, no solo por comportamiento real.

¿Se conserva, transforma o elimina? Se conserva.

Justificación: Es información legítima disponible en cualquier corte temporal, no un dato posterior al evento. Se documenta el riesgo, pero no amerita eliminarla.

TotalCharges

Cuándo se genera: Es un acumulado de facturación a la fecha.

¿Disponible al momento de predecir? Sí.

¿Podría inflar métricas artificialmente? Riesgo moderado por redundancia: correlación de 0.826 con tenure y 0.651 con MonthlyCharges. La diferencia frente a tenure × MonthlyCharges tiene media de 45.09, así que no es una copia exacta, pero comparte señal.

¿Se conserva, transforma o elimina? Se conserva.

Justificación: Aporta información adicional (variaciones de plan/promociones) más allá de lo que capturan tenure y MonthlyCharges por separado.

Decisión: Se conservan ambas variables ("tenure" y "TotalCharges"), que no representan información inexistente al momento de la predicción - son datos legítimos del historial del cliente disponibles en cualquier corte temporal. Se documenta esta ambigüedad conceptual como una limitación reconocida del modelo. Adicionalmente, se elimina "customerID" por no aportar valor predictivo
